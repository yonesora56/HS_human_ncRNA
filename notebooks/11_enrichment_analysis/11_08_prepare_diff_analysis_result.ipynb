## up5p Diff Analysis Result

```python
up5p_fantabio_diff_analysis_result = pl.read_parquet(
    "../../out/ChIP-Atlas_diff/diff_analysis_filter_parquet/transcript_cre/fantabio_HN2_up5p_v261002_diff_analysis_atac.parquet"
)

pl.Config.set_tbl_rows(2)
display(up5p_fantabio_diff_analysis_result)
```

&nbsp;

&nbsp;

## diff analysis + protein-coding transcripts

```python
up5p_fantabio_diff_analysis_result_pc = up5p_fantabio_diff_analysis_result.filter(
    pl.col("transcript_type") == "protein_coding"
).with_columns(
    # Drop the Ensembl version suffix, e.g. ENST00000791171.1 -> ENST00000791171
    pl.col("transcript_id").str.replace(r"\.\d+$", "")
)

up5p_fantabio_diff_analysis_result_pc.select(
    "transcript_id"
).unique().write_csv(
    "../../out/enrichment_analysis/query/up5p_fantabio_diff_analysis_atac_result_pc_transcript_list.txt",
    separator="\t",
    include_header=False
)

display(up5p_fantabio_diff_analysis_result_pc)
```
