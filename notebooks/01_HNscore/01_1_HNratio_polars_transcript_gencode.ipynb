```python
# import library
import os
import pathlib
import datetime
import numpy as np
import datetime as dt
import polars as pl
import polars.selectors as cs

from IPython.display import display
```

```python
VERSION = datetime.datetime.now().strftime("%y%m")
execute_date = datetime.datetime.now().strftime("%y%m%d")
print(f"Execute Date: {execute_date}")

# create directory
pathlib.Path(f"../../Data/HNscore/HNscore_transcript/HNscore_gencode_{VERSION}").mkdir(parents=True, exist_ok=True)
```

```python
# Upload gene expression data (quantitated by salmon v1.10.3)
# summarize transcript-level scaled TPM data (using pytximport v.0.12.0)
gene_exp_table = pl.read_csv(
    "../../Data/HNscore/HNscore_transcript/HNscore_gencode_2610/HS_human_gencode_v49_quant_transcriptlevel_20260930.tsv.gz", 
    separator='\t',
    n_threads=16
)

display(gene_exp_table)
```

```python
# One sum per sample column (string columns are skipped)
gene_exp_col_sum = gene_exp_table.select(cs.numeric()).sum()

display(gene_exp_col_sum)
```

```python
# Load metadata (bulk RNA-seq)
# ../metadata ディレクトリの notebook で処理、作成したメタデータを読み込む
HN_pair = pl.read_csv(
    f"../../Data/metadata/metadata_2610/rna_seq_all_metadata_simple_join_salmon_261001.tsv", 
    separator="\t",
    # truncate_ragged_lines=True,
    # infer_schema_length=100000,
    n_threads=16
)

display(HN_pair)
```

```python
# initialize HNratio table
HNratio = gene_exp_table.select(['transcript_id'])

# List to store failed pairs
failed_pairs = []

for row in HN_pair.to_dicts():
    h = row["sample_heat_id"] # H (heat-stress)
    n = row["sample_control_id"] # N (non-treatment)
        
    # Check column name exist
    h_exists = h in gene_exp_table.columns
    n_exists = n in gene_exp_table.columns
        
    log_message_2 = f"Stress column exists: {h_exists}, Control column exists: {n_exists}\n"
    print(log_message_2)  # Add print statement
        
    if h_exists and n_exists:
            
        # count the number of rows
        h_count = gene_exp_table[h].count()
        n_count = gene_exp_table[n].count()
        rows_match = h_count == n_count
            
            
        # check null value
        h_nulls = gene_exp_table[h].null_count()
        n_nulls = gene_exp_table[n].null_count()
            
        log_message_4 = f"Null values in Stress column: {h_nulls}, in Control column: {n_nulls}\n"
        print(log_message_4)  # Add print statement
            
        if rows_match and h_nulls == 0 and n_nulls == 0:
            ratio_column = ((gene_exp_table[h] + 1) / (gene_exp_table[n] + 1)).alias(h) # Not "gene_exp_table.select([h])" but "gene_exp_table[h]"
            HNratio = HNratio.with_columns(ratio_column)
            print(f"Successfully processed combination: {h} and {n}")  # Add success message
        else:
            print(f"Data integrity check failed. This combination {h} and {n} is skipped.")
            log_message_5 = f"Data integrity check failed. This combination {h} and {n} is skipped.\n"
            # Record failed pair with reason
            failed_pairs.append({
                "sample_heat_id": h,
                "sample_control_id": n,
                "failure_reason": f"Data integrity check failed: rows_match={rows_match}, h_nulls={h_nulls}, n_nulls={n_nulls}"
            })
    else:
        log_message_6 = "Error: One or both columns do not exist. This combination is skipped.\n"
        print(log_message_6)  # Add print statement
        # Record failed pair with reason
        failed_pairs.append({
            "sample_heat_id": h,
            "sample_control_id": n,
            "failure_reason": f"Column missing: h_exists={h_exists}, n_exists={n_exists}"
        })

# Convert failed pairs to DataFrame and save
if failed_pairs:
    failed_pairs_df = pl.DataFrame(failed_pairs)
    failed_pairs_df.write_csv(
        f"../../Data/HNscore/HNscore_transcript/HNscore_gencode_{VERSION}/failed_pairs_gencode_{execute_date}.tsv",
        separator="\t"
    )
    print(f"\nTotal failed pairs: {len(failed_pairs)}")
    print(f"Failed pairs saved to: failed_pairs_gencode_{execute_date}.tsv")
    display(failed_pairs_df)
else:
    pl.Config.set_fmt_str_lengths(200)
    print("\nAll pairs were successfully processed!")
```

```python
display(HNratio)
```

```python
# log2 transformation
HNratiolog2 = HNratio.select(
    [pl.col('transcript_id')] +
    [(pl.col(column).log(base=2)).alias(column) for column in HNratio.columns if column != 'transcript_id']
    )

display(HNratiolog2)
```
