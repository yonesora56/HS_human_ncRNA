## 1. HN-score Comparison

- 収集した全ての bulk RNA-seq データセット から算出した HN-score に基づく順位と、`stranded`と判断した bulk RNA-seq データセットのみから算出した HN-score に基づく順位の順位相関係数を算出する
- [scipy.stats.spearmanr](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.spearmanr.html)
- [scipy.stats.kendalltau](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.kendalltau.html)

```python
HN2_all_info_comparison = HN2_all_info.select(
    "transcript_id",
    "gene_id",
    "gene_symbol",
    "transcript_type",
    "HN2",
    "HN2_stranded"
).with_columns(
    (pl.col("HN2") - pl.col("HN2_stranded")).alias("HN2_non_stranded")
)

display(HN2_all_info_comparison)
```

```python
# Spearman correlation (ties get average ranks)
rho_all_vs_stranded, p_all_vs_stranded = stats.spearmanr(
    HN2_all_info_comparison["HN2"].to_numpy(),
    HN2_all_info_comparison["HN2_stranded"].to_numpy(),
)

# Independent comparison: stranded vs non-stranded datasets (no shared samples)
rho_stranded_vs_non_stranded, p_stranded_vs_non_stranded = stats.spearmanr(
    HN2_all_info_comparison["HN2_stranded"].to_numpy(),
    HN2_all_info_comparison["HN2_non_stranded"].to_numpy(),
)

rho_all_vs_non_stranded, p_all_vs_non_stranded = stats.spearmanr(
    HN2_all_info_comparison["HN2"].to_numpy(),
    HN2_all_info_comparison["HN2_non_stranded"].to_numpy(),
)

# Kendall tau-b as a tie-robust reference
tau_all_vs_stranded, p_tau_all_vs_stranded = stats.kendalltau(
    HN2_all_info_comparison["HN2"].to_numpy(),
    HN2_all_info_comparison["HN2_stranded"].to_numpy(),
)

tau_stranded_vs_non_stranded, p_tau_stranded_vs_non_stranded = stats.kendalltau(
    HN2_all_info_comparison["HN2_stranded"].to_numpy(),
    HN2_all_info_comparison["HN2_non_stranded"].to_numpy(),
)

tau_all_vs_non_stranded, p_tau_all_vs_non_stranded = stats.kendalltau(
    HN2_all_info_comparison["HN2"].to_numpy(),
    HN2_all_info_comparison["HN2_non_stranded"].to_numpy(),
)

# p-values below ~1e-308 underflow to 0.0 in float64
print(f"Spearman (all vs stranded): {rho_all_vs_stranded:.4f}, p = {p_all_vs_stranded:.4e}")
print(f"Spearman (stranded vs non-stranded): {rho_stranded_vs_non_stranded:.4f}, p = {p_stranded_vs_non_stranded:.4e}")
print(f"Spearman (all vs non-stranded): {rho_all_vs_non_stranded:.4f}, p = {p_all_vs_non_stranded:.4e}")
print(f"Kendall tau-b (all vs stranded): {tau_all_vs_stranded:.4f}, p = {p_tau_all_vs_stranded:.4e}")
print(f"Kendall tau-b (stranded vs non-stranded): {tau_stranded_vs_non_stranded:.4f}, p = {p_tau_stranded_vs_non_stranded:.4e}")
print(f"Kendall tau-b (all vs non-stranded): {tau_all_vs_non_stranded:.4f}, p = {p_tau_all_vs_non_stranded:.4e}")
```

```python
# weighted tau
# Larger scores are treated as more important (hyperbolic weight 1/(rank+1) by default)
# pvalue is always nan because the null distribution is unknown
HN2_all = HN2_all_info_comparison["HN2"].to_numpy()
HN2_stranded = HN2_all_info_comparison["HN2_stranded"].to_numpy()
HN2_non_stranded = HN2_all_info_comparison["HN2_non_stranded"].to_numpy()


# Emphasize the top (heat-induced) transcripts
wtau_up_all_vs_stranded, _ = stats.weightedtau(HN2_all, HN2_stranded)
wtau_up_stranded_vs_non_stranded, _ = stats.weightedtau(HN2_stranded, HN2_non_stranded)
wtau_up_all_vs_non_stranded, _ = stats.weightedtau(HN2_all, HN2_non_stranded)

# Emphasize the bottom (heat-repressed) transcripts by flipping the sign of both scores
wtau_down_all_vs_stranded, _ = stats.weightedtau(-HN2_all, -HN2_stranded)
wtau_down_stranded_vs_non_stranded, _ = stats.weightedtau(-HN2_stranded, -HN2_non_stranded)
wtau_down_all_vs_non_stranded, _ = stats.weightedtau(-HN2_all, -HN2_non_stranded)

print(f"Weighted tau, top-weighted (all vs stranded): {wtau_up_all_vs_stranded:.3f}")
print(f"Weighted tau, top-weighted (stranded vs non-stranded): {wtau_up_stranded_vs_non_stranded:.3f}")
print(f"Weighted tau, top-weighted (all vs non-stranded): {wtau_up_all_vs_non_stranded:.3f}")
print(f"Weighted tau, bottom-weighted (all vs stranded): {wtau_down_all_vs_stranded:.3f}")
print(f"Weighted tau, bottom-weighted (stranded vs non-stranded): {wtau_down_stranded_vs_non_stranded:.3f}")
print(f"Weighted tau, bottom-weighted (all vs non-stranded): {wtau_down_all_vs_non_stranded:.3f}")
```
