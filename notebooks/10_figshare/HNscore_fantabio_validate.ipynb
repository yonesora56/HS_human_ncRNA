## 1. all HNscore transcript - CRE hit pair relationships

- このデータを、`parquet` ファイルとして figshare に登録しておいて、ここに全部あることを示しておく
- 対応ついていない transcript は figshare に登録するデータは除いておく
- 大きすぎて figshare にアップロードできなかったため、こちらもカラムを落とす

```python
HNscore_transcript_all_fantabio = pl.read_parquet(
    f"../../out/fantabio/parquet_file/HNscore_transcript_all_fantabio_v261001.parquet"
).filter(
    pl.col("cre_id").is_not_null() # transcript - CRE pair のみにフィルタリング
)

display(HNscore_transcript_all_fantabio)
```

```python
HNscore_transcript_all_fantabio_figshare = HNscore_transcript_all_fantabio.drop(
    [
        "tags",
        "UP2",
        "DOWN2",
        "HGNC_ID",
        "UNCHANGED2",
        "UP2_stranded",
        "DOWN2_stranded",
        "UNCHANGED2_stranded",
        "number_of_tags",
        "polyA_features",
        "corrrespondence",
        "gene_source",
        "annotation_remarks",
        "annotation_remark_count",
        "tss_nearest_transcript_ids",
        "tss_nearest_transcript_distance",
        "screen_ccres",
        "chip_atlas_antigens",
        "silencerdbs",
        "fantom5_cage_peaks",
        "fantom5_enhancers",
        "itemRgb",
        "itemRgb_name"
    ]
)

display(HNscore_transcript_all_fantabio_figshare)
```

```python
# figshare に登録するファイル
HNscore_transcript_all_fantabio_figshare.write_csv(
    "../../out/figshare/HNscore_fantabio/HNscore_transcript_all_fantabio.tsv.gz",
    separator="\t",
    compression="gzip",
    compression_level=9
)

display(HNscore_transcript_all_fantabio_figshare)
```

## up5p transcript - CRE relationships

- 全部のアノテーションは parquet 形式のファイル に掲載し、それ以外のデータ、TSVファイルとして保存できるカラムを残し、figshare 用のファイルとする

```python
up5p_transcript_cre_pairs = pl.read_parquet(
    "../../out/fantabio/parquet_file/up5p_transcript_fantabio_v261001.parquet"
)

display(up5p_transcript_cre_pairs)
```

```python
up5p_transcript_cre_pairs_figshare = up5p_transcript_cre_pairs.drop(
    [
        "tags",
        "UP2",
        "DOWN2",
        "UNCHANGED2",
        "UP2_stranded",
        "DOWN2_stranded",
        "UNCHANGED2_stranded",
        "number_of_tags",
        "polyA_features",
        "corrrespondence",
        "gene_source",
        "annotation_remarks",
        "annotation_remark_count",
        "tss_nearest_transcript_ids",
        "tss_nearest_transcript_distance",
        "screen_ccres",
        "chip_atlas_antigens",
        "silencerdbs",
        "fantom5_cage_peaks",
        "fantom5_enhancers",
        "itemRgb",
        "itemRgb_name"
    ]
)

up5p_transcript_cre_pairs_figshare.write_csv(
    "../../out/figshare/HNscore_fantabio/up5p_transcript_cre_pairs.tsv",
    separator="\t"
)

display(up5p_transcript_cre_pairs_figshare)
```

## up5p transcript (no proximal CRE)

- fanta.bio version 1.2.0

```python
up5p_no_proximal_cre_transcripts = pl.read_parquet(
    "../../out/fantabio/no_proximal_cre_transcirpt_list/up5p_transcript_no_cre_v261001_validation.parquet"
)

display(up5p_no_proximal_cre_transcripts)
```

```python
up5p_no_proximal_cre_transcripts_figshare = up5p_no_proximal_cre_transcripts.select(
    "transcript_id",
    "seqname",
    "HN2",
    "HN2_stranded",
    "transcript_type",
    "gene_id",
    "gene_symbol",
    "cre_id",
    "cre_name",
    "fantabio_cre_class",
    "distance",
    "has_cre_gene",
    "flag"
).rename(
    {
        "cre_id": "cre_id_gene_level",
        "cre_name": "cre_name_gene_level",
        "fantabio_cre_class": "cre_class_gene_level",
        "distance": "distance_gene_level",
        "has_cre_gene": "has_cre_geneid",
    }
).sort(
    "HN2",
    descending=True
)

up5p_no_proximal_cre_transcripts_figshare.write_csv(
    "../../out/figshare/HNscore_fantabio/up5p_transcript_no_proximal_cre.tsv.gz",
    separator="\t",
    compression="gzip",
    compression_level=9
)

display(up5p_no_proximal_cre_transcripts_figshare)
```
