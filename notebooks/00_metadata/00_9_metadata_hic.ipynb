## GSE105028 (Hi-C, Hi-ChIP, ~~eU-seq,~~ ~~ChIP-seq~~, ~~ATAC-seq~~ )

- [Architectural Proteins and Pluripotency Factors Cooperate to Orchestrate the Transcriptional Response of hESCs to Temperature Stress](https://doi.org/10.1016/j.molcel.2018.07.012)
- [./12_7_metadata_proseq.ipynb](./12_7_metadata_proseq.ipynb)から処理を移動
- ATAC-seq のデータは、[./12_8_metadata_atacseq.ipynb](./12_8_metadata_atacseq.ipynb)に移動
- ChIP-seq のデータは、[./12_8_metadata_chipseq.ipynb](./12_8_metadata_chipseq.ipynb)に移動

```python
GSE105028_PEP = pl.read_csv(
    "../../Data/metadata/metadata_geofetch/GSE105028/GSE105028_PEP/GSE105028_PEP_raw.csv",
    separator=','
).filter(
    (pl.col("library_strategy") != "ATAC-Seq") &
    (pl.col("library_strategy") != "ChIP-Seq") &
    (pl.col("library_strategy") != "eU-Seq")
).sort(
    "srr",
    descending=False
).with_columns(
    pl.lit("https://doi.org/10.1016/j.molcel.2018.07.012").alias("reference_doi"), # add DOI of the reference paper column
    pl.lit("42 C").alias("stress treatment temperature"), # add heat treatment temperature column
    pl.lit("60 min.").alias("stress treatment time"), # add heat treatment time column (minutes)
    pl.lit("37 C (inferred)").alias("control treatment temperature") # add control temperature column
)

GSE105028_PEP.write_csv(
    "../../Data/metadata/metadata_hic/GSE105028_PEP_raw_filtered.tsv",
    separator='\t'
)

pl.Config.set_tbl_rows(2)
display(GSE105028_PEP)
```

```python
# Hi-C metadata
defined_pairs_gse105028_hic = [
    ("SRR6177928", "SRR6177926"), # hs_hic_rep1 vs. hic_rep1
    ("SRR6177929", "SRR6177927")  # hs_hic_rep2 vs. hic_rep2
]

paired_metadata_gse105028_hic = create_paired_metadata(
    GSE105028_PEP,
    defined_pairs_gse105028_hic,
    sample_id_column_name="srr"
)

if paired_metadata_gse105028_hic is not None:
    paired_metadata_gse105028_hic.write_csv(
        "../../Data/metadata/metadata_hic/GSE105028_hic_paired_metadata.tsv",
        separator='\t'
    )

pl.Config.set_tbl_rows(2)
display(paired_metadata_gse105028_hic)
```

```python
# Hi-ChIP metadata
defined_pairs_gse105028_hichip = [
    # RAD21
    ("SRR6206787", "SRR6206783"), # hs_rad21_hi-chip_rep1 vs. rad21_hi-chip_rep1
    ("SRR6206788", "SRR6206784"), # hs_rad21_hi-chip_rep2 vs. rad21_hi-chip_rep2
    ("SRR6206789", "SRR6206785"), # hs_rad21_hi-chip_rep3 vs. rad21_hi-chip_rep3
    ("SRR6206790", "SRR6206786"), # hs_rad21_hi-chip_rep4 vs. rad21_hi-chip_rep4
    # hs_rad21_hi-chip_rep1_and_rep2_technical_replicate (SRR6206791) と hs_ki_rad21_hi-chip (SRR6206792) は使用しない

    # CTCF
    ("SRR6206796", "SRR6206793"), # hs_ctcf_hi-chip_rep1 vs. ctcf_hi-chip_rep1
    ("SRR6206797", "SRR6206794"), # hs_ctcf_hi-chip_rep2 vs. ctcf_hi-chip_rep2
    # ctcf_hi-chip_rep3 (SRR6206795) は使用しない

    # OCT4
    ("SRR6206800", "SRR6206798"), # hs_oct4_hi-chip_rep1 vs. oct4_hi-chip_rep1
    ("SRR6206801", "SRR6206799"), # hs_oct4_hi-chip_rep2 vs. oct4_hi-chip_rep2

    # NANOG
    ("SRR6206804", "SRR6206802"), # hs_nanog_hi-chip_rep1 vs. nanog_hi-chip_rep1
    ("SRR6206805", "SRR6206803"), # hs_nanog_hi-chip_rep2 vs. nanog_hi-chip_rep2

    # KLF4 (replicate 2 がない)
    ("SRR6206808", "SRR6206806"), # hs_klf4_hi-chip vs. klf4_hi-chip_rep1
    ("SRR6206808", "SRR6206807"), # hs_klf4_hi-chip vs. klf4_hi-chip_rep2

    # H3K4me1
    ("SRR6206811", "SRR6206809"), # hs_h3k4me1_hi-chip_rep1 vs. h3k4me1_hi-chip_rep1
    ("SRR6206812", "SRR6206810"), # hs_h3k4me1_hi-chip_rep2 vs. h3k4me1_hi-chip_rep2

    # sinipbl_hs_ctcf_hi-chip (SRR6206813); sinipbl_hs_h3k4me1_hi-chip (SRR6206814) は使用しない

]

paired_metadata_gse105028_hichip = create_paired_metadata(
    GSE105028_PEP,
    defined_pairs_gse105028_hichip,
    sample_id_column_name="srr"
)

if paired_metadata_gse105028_hichip is not None:
    paired_metadata_gse105028_hichip.write_csv(
        "../../Data/metadata/metadata_hic/GSE105028_hichip_paired_metadata.tsv",
        separator='\t'
    )

pl.Config.set_tbl_rows(2)
display(paired_metadata_gse105028_hichip)
```

## Concat Dataframe

```python
df_hic_all: list[pl.DataFrame | None] = [
    paired_metadata_gse105028_hic,
    paired_metadata_gse105028_hichip,
    paired_metadata_gse130758_hic
]

dfs_hic:  list[pl.DataFrame] = [df for df in df_hic_all if df is not None]
```

```python
hic_concat = pl.concat(dfs_hic, how="diagonal")

hic_concat.write_csv(
    f"../../Data/metadata/metadata_{VERSION}/hic_metadata_all_{execute_date}.tsv",
    separator='\t'
)
```
