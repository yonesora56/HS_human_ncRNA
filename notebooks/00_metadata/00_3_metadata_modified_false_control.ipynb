## 1. GSE222110 (PAIRED) 

- ✅ DOWNLOAD
- ✅ TRIMMING
- ✅ QUANTIFICATION

&nbsp;

- validation date: 2025-05-04
- [An adaptive stress response that confers cellular resilience to decreased ubiquitination](https://doi.org/10.1038/s41467-023-43262-7)
- ユビキチンに関連する遺伝子のノックダウンに関連したデータの他、おそらく熱ストレスに関連したデータが含まれている
- 熱ショックは42℃で16時間の曝露？ (`Methods section` より)
- `E2combo siRNAs + heat shock` などが含まれている
- controlとなるデータの他、Heat shock (HS)となっているデータを収集する

```python
GSE222110_PEP = pl.read_csv(
    "../../Data/metadata/metadata_geofetch/GSE222110/GSE222110_PEP/GSE222110_PEP_raw.csv",
    separator=','
).filter(
    (pl.col("sample_library_strategy") == "RNA-Seq")
).filter(
    (pl.col("sample_title").str.contains(r'siRNA')) & 
    ~((pl.col("sample_title").str.contains(r'NTsiRNA_')) | 
      (pl.col("sample_title").str.contains(r'UBA1siRNA_'))
     )
).with_columns(
    (pl.col("biosample").str.split("/").list.get(-1).alias("sample_accession")),
    pl.lit("TruSeq stranded mRNA library prep kit (Illumina) (from GEO) / Illumina TruSeq RNA Sample Prep v2 Kit (from article)").alias("rna_library"), # add rna_library column from GEO page (Illumina)
    pl.lit("https://doi.org/10.1038/s41467-023-43262-7").alias("reference_doi"), # add DOI of the reference paper column
    pl.lit("42 C").alias("stress treatment temperature"), # add heat treatment temperature column
    pl.lit("16 h.").alias("stress treatment time"), # add heat treatment time column (minutes)
    pl.lit("37 C (inferred)").alias("control treatment temperature") # add control temperature column
).rename(
    {
        "sample_series_id": "gse_id",
        "big_key": "experiment_accession"
    }
).drop(
    "sra",
    "biosample"
)

# metadata from iseq
GSE222110 = pl.read_csv(
    "../../Data/metadata/metadata_iseq/GSE222110_metadata.tsv",
    separator='\t'
).drop(
    drop_columns_list
)

GSE222110_iseq_filtered = GSE222110.select([s.name for s in GSE222110 if not (s.null_count() == GSE222110.height)])

GSE222110_combined = GSE222110_PEP.join(
    GSE222110_iseq_filtered,
    on=["experiment_accession", "sample_accession"],
    how="left",
    coalesce=True
)

GSE222110_combined = GSE222110_combined[[s.name for s in GSE222110_combined if not (s.null_count() == GSE222110_combined.height)]]

GSE222110_combined.write_csv(
    "../../Data/metadata/metadata_modified/GSE222110_modified_metadata.tsv",
    separator='\t'
)

# display(GSE222110_combined)
```

```python
defined_pairs_gse222110 = [
    ("SRR22965783", "SRR22965789"), # CTRLsiRNA_HS_1 vs CTRLsiRNA_1
    ("SRR22965782", "SRR22965788"), # CTRLsiRNA_HS_2 vs CTRLsiRNA_2
    ("SRR22965781", "SRR22965787"), # CTRLsiRNA_HS_3 vs CTRLsiRNA_3
    ("SRR22965780", "SRR22965786"), # E2siRNAcombo_HS_1 vs E2siRNAcombo_1
    ("SRR22965779", "SRR22965785"), # E2siRNAcombo_HS_2 vs E2siRNAcombo_2
    ("SRR22965777", "SRR22965784")  # E2siRNAcombo_HS_3 vs E2siRNAcombo_2
]

paired_metadata_df_gse222110 = create_paired_metadata(
    combined_df=GSE222110_combined, 
    defined_sample_pairs=defined_pairs_gse222110,
    sample_id_column_name="accession" # default value, but specified to clarify
)

if paired_metadata_df_gse222110 is not None:
    paired_metadata_df_gse222110.write_csv(
        "../../Data/metadata/metadata_paired/GSE222110_paired_metadata.tsv",
        separator='\t'
    )

    display(paired_metadata_df_gse222110)
else:
    # Simple message helps you notice when no pairs were produced
    print("No paired metadata to write (create_paired_metadata returned None).")
```
