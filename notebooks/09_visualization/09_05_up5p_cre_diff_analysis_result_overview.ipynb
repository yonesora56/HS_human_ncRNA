## Common lncRNA

- transcript_id - cre_id ペアレベルで共通するかどうか check

```python
# Common lncRNA - CRE pairs: the same CRE is a differential peak region
# in both ATAC-seq and HSF1 ChIP-seq Diff Analysis.
# Diff Analysis columns exist in both tables, so suffix them with _atac / _hsf1.
DIFF_COLUMNS = ["log2fc", "q_value", "diff_dataset", "diff_dataset_count"]

common_lncrna_cre_pair = (
    up5p_fantabio_diff_analysis_result_lncrna
    .join(
        up5p_fantabio_diff_chipseq_hsf1_result_lncrna.select(
            "transcript_id", "cre_id", *DIFF_COLUMNS, "n_motif_hit"
        ),
        on=["transcript_id", "cre_id"],
        how="inner",
        suffix="_hsf1",
    )
    .rename({col: f"{col}_atac" for col in DIFF_COLUMNS})
    .sort("HN2", "transcript_id", "cre_id", descending=[True, False, False])
)

print(f"Common lncRNA - CRE pair (ATAC-seq & HSF1 ChIP-seq)  {common_lncrna_cre_pair.select(["transcript_id", "cre_id"]).n_unique()}")
print(f"Common lncRNA (ATAC-seq & HSF1 ChIP-seq)  {common_lncrna_cre_pair.select("transcript_id").n_unique()}")
print(f"Common CRE (ATAC-seq & HSF1 ChIP-seq)  {common_lncrna_cre_pair.select("cre_id").n_unique()}")
print(f"Common lncRNA - CRE pair with HSF1 motif  {common_lncrna_cre_pair.filter(pl.col("n_motif_hit").is_not_null()).height}")

pl.Config.set_tbl_rows(80)
display(
    common_lncrna_cre_pair.select(
        "transcript_id",
        "gene_id",
        "gene_symbol",
        "HN2",
        "cre_id",
        "fantabio_cre_class",
        "transcript_distance",
        "diff_dataset_count_atac",
        "diff_dataset_count_hsf1",
        "n_motif_hit",
    )
)
```

## Check _NEAT1_

- _NEAT1_ が、発現上昇転写産物群に含まれていないが、Diff Analysis では同定できているかどうかを check!
- そのためには、全ての transcript - CRE のリストと、up5p transcripts と統合する前の diff analysis で同定された CRE list が必要っちゅうわけだな...

```python
HNscore_all_human_transcript_cre_pair = pl.read_parquet(
    "../../out/fantabio/parquet_file/HNscore_transcript_all_fantabio_v261001.parquet"
)

NEAT1_cre_pair = (
    HNscore_all_human_transcript_cre_pair
    .filter(
        pl.col("gene_symbol") == "NEAT1"
    )
)

pl.Config.set_tbl_rows(2)
display(NEAT1_cre_pair)
```

```python
# ATAC-seq DPR CREs
atacseq_dpr_cre = pl.read_parquet(
    "../../out/ChIP-Atlas_diff/diff_analysis_filter_parquet/only_cre/fantabio_cre_list_diff_analysis_atac.parquet"
)

neat1_dpr_cre = (
    atacseq_dpr_cre
    .filter(
        # Keep DPR CREs that are paired with any NEAT1 transcript
        pl.col("cre_id").is_in(NEAT1_cre_pair["cre_id"].implode())
    )
)

display(neat1_dpr_cre)
```

```python
# ChIP-seq DPR CREs (HSF1)
chipseq_hsf1_dpr_cre = pl.read_parquet(
    "../../out/ChIP-Atlas_diff/diff_analysis_filter_parquet/only_cre/fantabio_cre_list_diff_analysis_chipseq_hsf1.parquet"
)

neat1_chipseq_hsf1_dpr_cre = (
    chipseq_hsf1_dpr_cre
    .filter(
        # Keep DPR CREs that are paired with any NEAT1 transcript
        pl.col("cre_id").is_in(NEAT1_cre_pair["cre_id"].implode())
    )
)

display(neat1_chipseq_hsf1_dpr_cre)
```
