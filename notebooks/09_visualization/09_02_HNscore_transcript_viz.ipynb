## Select

- tag 情報を用いるため、全ての HN-score 表から、選抜する

```python
up5p_complete = HNscore_transcript.filter(
    # Fold the Series into one scalar list before the membership test.
    pl.col("transcript_id").is_in(up5p["transcript_id"].implode())
)

down5p_complete = HNscore_transcript.filter(
    pl.col("transcript_id").is_in(down5p["transcript_id"].implode())
)

# Each source id is unique, so equal height means every row was recovered.
print(f"up5p rows: complete={up5p_complete.height}, source={up5p.height}")
assert up5p_complete.height == up5p.height

print(f"down5p rows: complete={down5p_complete.height}, source={down5p.height}")
assert down5p_complete.height == down5p.height


display(up5p_complete)
display(down5p_complete)
```

## Scatter Plots

- For Figure

```python
# Rank every transcript by HN2. Red and blue are the lists selected above.
threshold = 2
count = HNscore_transcript.sort(f"HN{threshold}", descending=True)

# The saved lists must be the same tails as a fresh 5% cutoff, including ties.
n_rows = count.height
tail_n = int(n_rows * 0.05)
hn_score_series = count.get_column(f"HN{threshold}")
top_5_percent_threshold = hn_score_series.top_k(tail_n).min()
bottom_5_percent_threshold = hn_score_series.bottom_k(tail_n).max()

threshold_up_ids = (
    count.filter(pl.col(f"HN{threshold}") >= top_5_percent_threshold)
    .select("transcript_id")
    .sort("transcript_id")
)

threshold_down_ids = (
    count.filter(pl.col(f"HN{threshold}") <= bottom_5_percent_threshold)
    .select("transcript_id")
    .sort("transcript_id")
)
```

```python
# 既に選抜した transcript_id list と同じかどうかを確認
assert up5p_complete.select("transcript_id").sort("transcript_id").equals(threshold_up_ids)
assert down5p_complete.select("transcript_id").sort("transcript_id").equals(threshold_down_ids)
```

```python
# Row number after the HN2 sort is the x-axis rank.
ranked = count.with_row_index("rank")
hn_col = f"HN{threshold}"
x_data = ranked.get_column("rank").to_numpy()
y_data = ranked.get_column(hn_col).to_numpy()

upregulated = ranked.filter(
    pl.col("transcript_id").is_in(up5p_complete.get_column("transcript_id").implode())
)
upregulated_indices = upregulated.get_column("rank").to_numpy()
y_data_up = upregulated.get_column(hn_col).to_numpy()

downregulated = ranked.filter(
    pl.col("transcript_id").is_in(down5p_complete.get_column("transcript_id").implode())
)
downregulated_indices = downregulated.get_column("rank").to_numpy()
y_data_down = downregulated.get_column(hn_col).to_numpy()
```

```python
# ggplot supplies the light grid; white face matches the figure.
plt.style.use("ggplot")
plt.rcParams["figure.figsize"] = [10, 16]

ax = plt.axes()
ax.set_facecolor('white')

#All genes
plt.scatter(
            x_data,
            y_data,
            s = 30,
            c="grey",
            edgecolors="black",
            linewidths=0.01,
            alpha=0.5,
            label="all genes"
            )

# upregulated genes
plt.scatter(
            upregulated_indices,
            y_data_up,
            s=100,
            c="red",
            edgecolors="black",
            linewidths=0.2,
            alpha=1,
            label="upregulated"
            )

#downregulated genes
plt.scatter(
            downregulated_indices,
            y_data_down,
            s=100,
            c="blue",
            edgecolors="white",
            linewidths=0.2,
            alpha=1,
            label="downregulated"
            )

plt.grid(color='b', linestyle=':', linewidth=0.1)
plt.xlabel("Transcript ranking based on HN-score (GENCODE release49)", fontsize=18) 
plt.ylabel("HN-score (human) ", fontsize=18)
plt.tick_params(labelsize=12)

plt.savefig(
    "../../out/main_figure/HNscore_all_transcript_scatter_plot.png",
    dpi=500,
    bbox_inches="tight"
)
plt.show()
```

## Assign Original Flag

- `protein_coding`, `lncRNA`, `other` の 3つに大きく分ける

```python
up5p_complete_classification = (
    up5p_complete
    .with_columns(
        pl.when(pl.col("transcript_type") == "protein_coding")
        .then(pl.lit("protein_coding"))
        .when(pl.col("transcript_type") == "lncRNA")
        .then(pl.lit("lncRNA"))
        .otherwise(pl.lit("others"))
        .alias("original_flag")
    )
)

down5p_complete_classification = (
    down5p_complete
    .with_columns(
        pl.when(pl.col("transcript_type") == "protein_coding")
        .then(pl.lit("protein_coding"))
        .when(pl.col("transcript_type") == "lncRNA")
        .then(pl.lit("lncRNA"))
        .otherwise(pl.lit("others"))
        .alias("original_flag")
    )
)

display(up5p_complete_classification)
display(down5p_complete_classification)
```

### up5p

```python
# Define custom colors for each gene category
transcript_category_colors = {
    'protein_coding': '#EF553B',  # Red-orange (画像のmRNAに合わせて赤みがかったオレンジ色)
    'others': '#D3D3D3',  # LightGray (薄い灰色)
    'lncRNA': '#AB63FA'  # Purple (画像のDivergent p-lncRNAに合わせて紫色)
    # 'p-lncRNA, intergenic': '#4682B4',  # Steelblue (画像のIntergenic p-lncRNAに合わせてsteelblue)
    # 'e-lncRNA': '#00CC96'  # Teal-green (画像のe-lncRNAに合わせて緑色)
}

transcript_category_counts_up = (
    up5p_complete_classification
    .group_by("original_flag")  # Group by gene category column
    .agg(pl.len().alias("count"))  # Count occurrences
    .sort("count", descending=True)  # Sort by count in descending order
)

# Enhanced pie chart with custom colors
fig_basic = px.pie(
    transcript_category_counts_up.to_pandas(),  # Convert Polars to Pandas for plotly
    values='count',
    names='original_flag',
    title='Transcript Type Distribution (Upregulated genes)',
    color='original_flag',  # Add color parameter
    color_discrete_map=transcript_category_colors  # Apply custom color mapping
)

# Customize the appearance
fig_basic.update_traces(
    textposition='outside',  # Place text outside pie slices
    textinfo='value+percent+label',  # Show value, percentage and label
    hovertemplate='<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}<extra></extra>'
)

# Update layout for better appearance
fig_basic.update_layout(
    font_size=18,
    showlegend=True,
    legend=dict(
        orientation="v",
        yanchor="middle",
        y=0.5,
        xanchor="left",
        x=1.01
    )
)

fig_basic.show()
```

## Pie Chart (lncRNA annotation remarks)

- 新たに remark をつける

```python
top5_genes_lncRNA = (
    up5p_complete.filter(
        pl.col("transcript_type") == "lncRNA"
    )
    .with_columns(
        # when() is only a builder; then/otherwise makes the Expr with_columns accepts.
        pl
        .when(pl.col("HGNC_ID").is_not_null() & pl.col("annotation_remarks").list.eval(pl.element().str.contains("CLS3 project")).list.any() & pl.col("tags").list.eval(pl.element().str.contains("TAGENE")).list.any())
        .then(pl.lit("Gene Symbol + CLS3 project"))

        .when(pl.col("HGNC_ID").is_null() & pl.col("annotation_remarks").list.eval(pl.element().str.contains("CLS3 project")).list.any())
        .then(pl.lit("No Gene Symbol + CLS3 project"))

        .when(pl.col("HGNC_ID").is_not_null() & pl.col("tags").list.eval(pl.element().str.contains("basic")).list.any())
        .then(pl.lit("Gene Symbol + basic"))

        .when(pl.col("tags").list.eval(pl.element().str.contains("basic")).list.any())
        .then(pl.lit("basic"))

        .otherwise(pl.lit("Other"))
        .alias("representative_annotation_remarks")
    )
)

print(f"top5 lncRNA transcript count: {top5_genes_lncRNA.select('transcript_id').n_unique()}")
display(top5_genes_lncRNA)
```

```python
# check
top5_genes_lncRNA_other = top5_genes_lncRNA.filter(pl.col("representative_annotation_remarks") == "Other")
display(top5_genes_lncRNA_other)
```

```python
# Define custom colors for each gene category
annotation_remark_colors = {
    'Gene Symbol + CLS3 project': '#EF553B',  # Red-orange (画像のmRNAに合わせて赤みがかったオレンジ色)
    'Other': '#D3D3D3',  # LightGray (薄い灰色)
    'No Gene Symbol + CLS3 project': '#AB63FA',  # Purple (画像のDivergent p-lncRNAに合わせて紫色)
    'Gene Symbol + basic': '#4682B4',  # Steelblue (画像のIntergenic p-lncRNAに合わせてsteelblue)
    'basic': '#00CC96'  # Teal-green (画像のe-lncRNAに合わせて緑色)
}

transcript_category_counts_up = (
    top5_genes_lncRNA
    .group_by("representative_annotation_remarks")  # Group by gene category column
    .agg(pl.len().alias("count"))  # Count occurrences
    .sort("count", descending=True)  # Sort by count in descending order
)

# Enhanced pie chart with custom colors
fig_basic = px.pie(
    transcript_category_counts_up.to_pandas(),  # Convert Polars to Pandas for plotly
    values='count',
    names='representative_annotation_remarks',
    title='LncRNA Representative Annotation Remarks Distribution (Upregulated genes)',
    color='representative_annotation_remarks',  # Add color parameter
    color_discrete_map=annotation_remark_colors  # Apply custom color mapping
)

# Customize the appearance
fig_basic.update_traces(
    textposition='outside',  # Place text outside pie slices
    textinfo='value+percent+label',  # Show value, percentage and label
    hovertemplate='<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}<extra></extra>',
    # Inset the pie so long labels sit in the margin instead of on each other.
    domain=dict(x=[0.28, 0.72], y=[0.08, 0.92]),
)

# Update layout for better appearance.
# The slice labels already name each category, so a side legend would cover them.
fig_basic.update_layout(
    font_size=15,
    showlegend=False,
    width=1000,
    height=640,
    margin=dict(l=10, r=10, t=70, b=30),
)

fig_basic.show()
```
