## `fanta.bio` resource (Tags per million (TPM) data)

- 熱ストレス下のCAGEデータと比較できるかもしれないので、読み込んでおく
- 栂さんから教えてもらった[NET-CAGEの論文](https://doi.org/10.1038/s41588-019-0485-9)のFig.6に記載がある

```python
tpm_path = "../../Data/Database/fantabio/human-CREv1.2.1.hg38.cre-peaks.tpm.txt.gz"
# Skip ## comment lines. read_csv n_rows counts those lines, so n_rows=1 returns no data.
with gzip.open(tpm_path, "rt") as handle:
    tpm_columns = next(
        csv.reader(
            (line for line in handle if not line.startswith("##")),
            delimiter="\t",
        )
    )

# Sample columns are TPM. Lock them to Float64 so integer-looking rows are not inferred as Int64.
fantabio_tpm = pl.read_csv(
    tpm_path,
    separator="\t",
    n_threads=16,
    comment_prefix="##",
    schema={
        name: pl.String if name == "creRegion" else pl.Float64
        for name in tpm_columns
    },
).with_columns(
    # "creRegion" カラムの内容を分割して、最初の要素 (fanta.bio の CRE ID) を "cre_id" カラムとして抽出
    pl.col(
        "creRegion"
    ).str.split(
        "|"
    ).list.first().alias(
        "creRegion"
    )
).rename(
    {
        "creRegion": "cre_id"
    }
)

display(fantabio_tpm)
```

## 全サンプルへの拡張：群別平均 TPM → entropy → expression specificity

上の **THP-1 の例と同じく、対象の `Source Name` の TPM を直接平均する**。
元の17セルは変更していない。`fantabio_tpm` は線形値の Polars DataFrame として既に読み込まれている。
追加部分では、大きい中間表を LazyFrame とし、`head(5).collect()` で表示する。

**何を同じにするか。** 群内の算術平均は添付テンプレートが基準。
以前の回答にあった「BioSample 内を平均 → 群内を平均」という二段階集約は行わない。
同じ BioSample の測定数が不均等なら、両者の値は異なる。

今回の全群平均は、処理条件や `Protocol Name` を自動分割しない探索的な値。
異なる測定法や刺激・分化条件が含まれても、そのままの群平均であり、未処理試料だけの特異性ではない。
注釈のない試料を一つの「unknown 群」にまとめたり、同じ試料を複数群に二重計上したりはしない。

### 1. サンプルと注釈を、見やすい列名の表にする

THP-1 の元の列順を保つため、既存の cell 表 → tissue 表 → 残りの metadata の順で並べる。
重複して登場した同じ `Source Name` は一度だけ使う。元の metadata 自体に重複 ID があれば停止する。

```python
assert fantabio_sample_metadata["Source Name"].null_count() == 0
assert fantabio_sample_metadata["Source Name"].n_unique() == fantabio_sample_metadata.height

fantabio_sample_ordered = pl.concat([
    fantabio_sample_cell, fantabio_sample_uberon, fantabio_sample_metadata,
]).unique(subset="Source Name", keep="first", maintain_order=True)

display(fantabio_sample_ordered.head(5))
```

```python
fantabio_sample_annotation = fantabio_sample_ordered.select(
    "Source Name", "Comment [biosample_id]", "Protocol Name",
    "Characteristics [sample type]", "Characteristics [description]",
    pl.col("Term Accession Number_duplicated_0").alias("cell_line_id"),
    pl.col("Term Accession Number").alias("cell_id"),
    pl.col("Term Accession Number_duplicated_2").alias("tissue_id"),
    pl.col("Term Accession Number_duplicated_1").alias("efo_id"),
    pl.col("Characteristics [cell line]").alias("cell_line_name"),
    pl.col("Characteristics [CL term]").alias("cell_name"),
    pl.col("Characteristics [tissue]").alias("tissue_name"),
    pl.col("Characteristics [EFO term]").alias("efo_name"),
).with_columns(
    pl.col("cell_line_id", "cell_id", "tissue_id", "efo_id")
    .str.strip_chars().replace("", None)
)

display(fantabio_sample_annotation.head(5))
```

### 2. 各試料を一つの集約群に割り当てる

ID の優先順位は **Cellosaurus → CL → UBERON → EFO**。
例えば細胞株と由来組織の両方が注釈されていても、細胞株の群に一度だけ入れる。
これは今回明示的に選んだ分類規則であり、ontology 間の同義関係を仮定したものではない。
CL の注釈があることだけで primary cell と判定はしない。

```python
fantabio_sample_assignment = fantabio_sample_annotation.with_columns(
    pl.coalesce("cell_line_id", "cell_id", "tissue_id", "efo_id").alias("group_id"),
).with_columns(
    pl.when(pl.col("cell_line_id").is_not_null()).then(pl.lit("cell_line"))
    .when(pl.col("cell_id").is_not_null()).then(pl.lit("cell"))
    .when(pl.col("tissue_id").is_not_null()).then(pl.lit("tissue"))
    .when(pl.col("efo_id").is_not_null()).then(pl.lit("EFO_only"))
    .otherwise(None).alias("group_type"),
    pl.when(pl.col("cell_line_id").is_not_null()).then(pl.col("cell_line_name"))
    .when(pl.col("cell_id").is_not_null()).then(pl.col("cell_name"))
    .when(pl.col("tissue_id").is_not_null()).then(pl.col("tissue_name"))
    .otherwise(pl.col("efo_name")).fill_null(pl.col("group_id")).alias("group_name"),
    pl.col("Source Name").is_in(fantabio_tpm.columns).alias("in_tpm"),
)

display(fantabio_sample_assignment.select(
    "Source Name", "group_id", "group_name", "group_type", "Protocol Name", "in_tpm"
).head(10))
```

### 3. 対応しないサンプルを表示する

「全サンプル」は、matrix に存在し、上のいずれかの ID に割り当てられる全試料を指す。
注釈なしの試料は `fantabio_unassigned_samples` に残す。metadata と matrix の版違いは黙って無視せず停止する。

```python
fantabio_matrix_samples = pl.DataFrame({"Source Name": fantabio_tpm.columns[1:]})
fantabio_missing_metadata = fantabio_matrix_samples.join(
    fantabio_sample_assignment.select("Source Name"), on="Source Name", how="anti"
)
fantabio_missing_tpm = fantabio_sample_assignment.filter(~pl.col("in_tpm"))
fantabio_unassigned_samples = fantabio_sample_assignment.filter(pl.col("group_id").is_null())

display(fantabio_missing_metadata.head(10))
display(fantabio_missing_tpm.head(10))
display(fantabio_unassigned_samples.head(10))
print("注釈のない試料数:", fantabio_unassigned_samples.height)
assert fantabio_missing_metadata.is_empty(), "TPM にある試料が metadata にありません。"
assert fantabio_missing_tpm.is_empty(), "metadata にある試料が TPM にありません。"
```

```python
fantabio_sample_used = fantabio_sample_assignment.filter(pl.col("group_id").is_not_null())
assert not fantabio_sample_used.is_empty(), "集約可能な試料がありません。"

display(fantabio_sample_used.select(
    "Source Name", "group_id", "group_name", "group_type", "Protocol Name"
).head(10))
```

### 4. 「群 ID → 対象サンプル列」の対応表を作る

`source_names` は Polars の List 列。THP-1 の `THP1_cell_source_names` に相当するリストを全群について保持する。
`n_samples` はライブラリー数であり、独立な donor 数ではない。
同じ ID に複数のラベルがある場合も、`group_names` にすべて残す。

```python
fantabio_group_metadata = fantabio_sample_used.group_by("group_id", maintain_order=True).agg(
    pl.col("group_name").unique(maintain_order=True).alias("group_names"),
    pl.col("group_type").first(),
    pl.col("Source Name").alias("source_names"),
    pl.len().alias("n_samples"),
    pl.col("Protocol Name").unique().sort().alias("protocols"),
    pl.col("Characteristics [sample type]").unique().sort().alias("sample_types"),
).with_columns(
    (pl.lit("mean_tpm_") + pl.col("group_id")).alias("mean_column"),
    pl.col("protocols").list.len().alias("n_protocols"),
)

display(fantabio_group_metadata.head(10))
display(fantabio_group_metadata.filter(pl.col("n_protocols") > 1).head(10))
print("集約群数:", fantabio_group_metadata.height)
```

### 5. 入力 TPM の確認

ここだけは、注釈に使う全サンプル・全 CRE を検査する。巨大な値の表を long format に変換するのではなく、
サンプルごとの異常値の件数だけを小さい表にする。
`null`・NaN・Inf・負値があれば停止し、ゼロへの置換や無視はしない。
元の THP-1 平均と同じ結果を検証する対象は、この検査を通る入力である。

```python
used_source_names = fantabio_sample_used["Source Name"].to_list()
fantabio_tpm_qc = (
    fantabio_tpm.lazy().select(used_source_names)
    .select((pl.all().is_null() | ~pl.all().is_finite() | (pl.all() < 0)).sum())
    .collect()
    .unpivot(variable_name="Source Name", value_name="n_invalid")
)

display(fantabio_tpm_qc.filter(pl.col("n_invalid") > 0))
assert fantabio_tpm_qc["n_invalid"].sum() == 0, "無効な TPM が含まれています。"
assert fantabio_tpm["cre_id"].null_count() == 0
assert fantabio_tpm["cre_id"].n_unique() == fantabio_tpm.height
```

### 6. 全群の平均 TPM を横に並べる

THP-1 の `pl.mean_horizontal(source_names)` を、全群について同じように指定する。
下の `for` は**小さい metadata 表から列の計算式を並べるだけ**で、CRE の数値を Python のループで計算しない。
元の表の行順が共通なので、群ごとに `join` を繰り返す代わりに、一度の `select` で横に並べられる。

```python
fantabio_tpm_mean = fantabio_tpm.lazy().select(
    "cre_id",
    *[
        pl.mean_horizontal(source_names).alias(mean_column)
        for mean_column, source_names in fantabio_group_metadata.select(
            "mean_column", "source_names"
        ).iter_rows()
    ],
)

display(fantabio_tpm_mean.head(5).collect())
```

### 7. 元の THP-1 の計算との一致を確認する

最初に、EFO 名で選んだ元の試料群と、Cellosaurus ID で選んだ群が同じ試料集合か確認する。
違っていれば、計算ライブラリーではなく**対象試料の違い**なので停止する。
一致すれば全 CRE の THP-1 平均を元の DataFrame と比較する。入力値や集約順序を変更していないため、ここは完全一致で検査する。

```python
from polars.testing import assert_frame_equal

thp1_mean_column = fantabio_tpm_thp1.columns[1]
thp1_sources_all = (
    fantabio_group_metadata.filter(pl.col("mean_column") == thp1_mean_column)
    .select("source_names").explode("source_names")["source_names"].to_list()
)
assert sorted(thp1_sources_all) == sorted(THP1_cell_source_names), "THP-1 の対象試料が違います。"

fantabio_thp1_check = fantabio_tpm_mean.select("cre_id", thp1_mean_column).collect()
assert_frame_equal(fantabio_thp1_check, fantabio_tpm_thp1, check_exact=True)
display(fantabio_thp1_check.head(5))
print("THP-1: 対象試料と全 CRE の平均 TPM がテンプレートと一致しました。")
```

### 8. 特異性の比較対象を選ぶ

既定値は、上で作成した **全種類の注釈付き biosample 群**。
この混合集合での値を「純粋な組織特異性」とは解釈しない。
組織だけの解析にする場合は下のコメント行に置き換える。その場合、群数と比較対象が変わるためスコアも変わる。

```python
specificity_groups = fantabio_group_metadata
# 組織のみを比較する場合:
# specificity_groups = fantabio_group_metadata.filter(pl.col("group_type") == "tissue")

mean_tpm_columns = specificity_groups["mean_column"].to_list()
assert len(mean_tpm_columns) >= 2, "特異性には2群以上が必要です。"
display(specificity_groups.select("group_id", "group_names", "group_type", "n_samples"))
```

### 9. 各 CRE の群別平均を、一つの List 列にまとめる

`group_tpm` は Python のリストへ変換したものではなく、**Polars の List 型の列**。
リストの要素の順番は `specificity_groups` の行順と対応する。
List にすることで、元の巨大な行列を `unpivot` せず、行ごとに entropy を計算できる。

```python
fantabio_group_values = fantabio_tpm_mean.select(
    "cre_id", pl.concat_list(mean_tpm_columns).alias("group_tpm"),
)

display(fantabio_group_values.head(5).collect())
```

```python
fantabio_group_values = fantabio_group_values.with_columns(
    pl.col("group_tpm").list.len().alias("n_groups"),
    pl.col("group_tpm").list.sum().alias("sum_group_tpm"),
    pl.col("group_tpm").list.max().alias("max_group_tpm"),
)

display(fantabio_group_values.head(5).collect())
```

### 10. Polars の `Expr.entropy` を使う

`list.eval` の内側は CRE **一行分**の群別 TPM。`normalize=True` により群間の比率へ正規化する。
自然対数の entropy を求める。`value_counts` やヒストグラムから計算するのではない。

ゼロは $0\log0=0$ なので entropy の内部では正の要素だけを渡す。
**`n_groups` はゼロの群を含む元の長さのまま**。全群ゼロは分布が定義できないため `null` にする。
ここで使うのは empirical Shannon entropy であり、Chao–Shen 補正ではない。

```python
fantabio_entropy = fantabio_group_values.with_columns(
    pl.when(pl.col("sum_group_tpm") > 0).then(
        pl.col("group_tpm").list.eval(
            pl.element().filter(pl.element() > 0).entropy(normalize=True)
        ).list.first()
    ).otherwise(None).alias("entropy"),
)

display(fantabio_entropy.head(5).collect())
```

### 11. Expression specificity を計算する

$$S=1-\frac{H}{\ln K}$$

$K$ は `n_groups`。全群同じ値なら $S=0$、一群のみ正なら $S=1$。
全群ゼロは `null` のままにする。`clip` は丸め誤差による僅かな範囲外の値を防ぐ。
pseudo-count、log2 変換、TPM の検出閾値による切り捨ては行わない。

```python
fantabio_specificity = fantabio_entropy.with_columns(
    (1 - pl.col("entropy") / pl.col("n_groups").cast(pl.Float64).log())
    .clip(0.0, 1.0).alias("specificity"),
).drop("group_tpm")

display(fantabio_specificity.head(10).collect())
```

### 12. 既存の CRE 注釈を結合する

結果の値を変える操作ではなく、`cre_id` による注釈の追加。
`validate="1:1"` によって重複キーによる行の増加を防ぐ。

```python
fantabio_specificity_bed = fantabio_specificity.join(
    fantabio_bed_select.lazy(), on="cre_id", how="left", validate="1:1",
)
fantabio_specificity_pla = fantabio_specificity_bed.filter(pl.col("fantabio_cre_class") == "PLA")
fantabio_specificity_ela = fantabio_specificity_bed.filter(pl.col("fantabio_cre_class") == "ELA")

display(fantabio_specificity_bed.head(5).collect())
```

## 任意の照合：SciPy と同じ定義になっているか

**この節は本計算とは独立した確認用**。この節だけで SciPy を使い、先頭1,000 CRE の小さい配列を作る。
主解析には pandas、NumPy、SciPy、Python の独自集約関数を使用しない。

同じ非負の群別平均を使う限り、両ライブラリーは同じ Shannon entropy を定義する。
実装の加算順序による微小な丸め差はあり得るため、ここは絶対誤差 `1e-12` で照合する。
全群ゼロは SciPy の NaN を `null` に変えて、今回の未定義値の方針にそろえる。

```python
from scipy.stats import entropy as scipy_entropy

entropy_input_check = fantabio_tpm_mean.select("cre_id", *mean_tpm_columns).head(1000).collect()
entropy_scipy_check = entropy_input_check.select("cre_id").with_columns(
    pl.Series("entropy_scipy", scipy_entropy(
        entropy_input_check.select(mean_tpm_columns).to_numpy(), axis=1,
    )).fill_nan(None),
)

entropy_comparison = fantabio_entropy.head(1000).collect().join(
    entropy_scipy_check, on="cre_id", how="left", validate="1:1",
).select("cre_id", "n_groups", "sum_group_tpm", "entropy", "entropy_scipy")
display(entropy_comparison.head(10))
```

```python
entropy_comparison = entropy_comparison.with_columns(
    (pl.col("entropy") - pl.col("entropy_scipy")).abs().alias("absolute_error"),
).with_columns(
    ((pl.col("entropy").is_null() & pl.col("entropy_scipy").is_null()) |
     (pl.col("absolute_error") <= 1e-12)).fill_null(False).alias("match"),
)

display(entropy_comparison.select(
    pl.len().alias("n_checked"), pl.col("absolute_error").max(), pl.col("match").all()
))
assert entropy_comparison["match"].all(), "SciPy との不一致を確認してください。"
```
